# Hotel_Booking

```python

# 1. Libraries import karna
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Dataset read karna
data = pd.read_csv("hotel_booking.csv")

# 3. Dataset display karna
print(data.head())

# 4. Dataset information
print(data.info())

# 5. Missing values check karna
print(data.isnull().sum())

# 6. Missing values handle karna
data = data.fillna(0)

# 7. Duplicate values check karna
print(data.duplicated().sum())

# 8. Duplicate values remove karna
data = data.drop_duplicates()

# 9. Statistical analysis
print(data.describe())

# 10. Visualization

# Hotel bookings count
sns.countplot(x="hotel", data=data)
plt.show()

# Cancellations count
sns.countplot(x="is_canceled", data=data)
plt.show()

# Average Daily Rate histogram
plt.hist(data["adr"])
plt.xlabel("Average Daily Rate")
plt.ylabel("Count")
plt.show()

# 11. Clean dataset save karna
data.to_csv("cleaned_hotel_booking.csv", index=False)

print("Dataset completed successfully!")
```
