**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_roll_number(self):
        return self.__roll_number

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


s1 = Student("Fatima", 101)
print(s1.get_name())
print(s1.get_roll_number())

s1.set_name("Ayesha")
s1.set_roll_number(102)
print(s1.get_name())
print(s1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Enter a valid amount")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance or invalid amount")

    def show_balance(self):
        print("Balance:", self.__balance)


account = BankAccount(5000)
account.deposit(1000)
account.withdraw(2000)
account.show_balance()
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self, salary):
        self.__salary = 0
        self.set_salary(salary)

    def set_salary(self, salary):
        if salary < 0:
            print("Invalid Salary")
        else:
            self.__salary = salary

    def get_salary(self):
        return self.__salary


emp = Employee(30000)
print("Salary:", emp.get_salary())
emp.set_salary(-5000)
print("Salary after trying negative value:", emp.get_salary())
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
class Example:
    def __init__(self):
        self._protected = "This is protected"
        self.__private = "This is private"

    def show_private(self):
        print(self.__private)


obj = Example()
print(obj._protected)  # Protected variable can be accessed, but it is meant for internal use
obj.show_private()
# obj.__private would give an AttributeError because of name mangling
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

```python
class Account:
    def __init__(self, pin):
        self.__pin = pin

    def check_pin(self):
        print("PIN is protected inside the class")


a1 = Account(1234)
a1.check_pin()

try:
    print(a1.__pin)
except AttributeError as error:
    print("Error:", error)

# Encapsulation helps keep important data controlled and prevents direct changes.
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model


my_car = Car("Toyota", "Corolla")
print("Brand:", my_car.brand)
print("Model:", my_car.model)
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
class Animal:
    def eat(self):
        print("Animal eats food")


class Mammal(Animal):
    def walk(self):
        print("Mammals can walk")


class Dog(Mammal):
    def bark(self):
        print("Dog says woof")


my_dog = Dog()
my_dog.eat()
my_dog.walk()
my_dog.bark()
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
class Shape:
    def show(self):
        print("This is a shape")


class Circle(Shape):
    def draw(self):
        print("Drawing a circle")


class Square(Shape):
    def draw(self):
        print("Drawing a square")


circle = Circle()
square = Square()
circle.show()
circle.draw()
square.show()
square.draw()

# Circle and Square both inherit from the same parent class Shape.
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
class Bird:
    def fly(self):
        print("Most birds can fly")


class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly")


bird = Bird()
ostrich = Ostrich()
bird.fly()
ostrich.fly()
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
class Dog:
    def sound(self):
        print("Dog says woof")


class Cat:
    def sound(self):
        print("Cat says meow")


class Cow:
    def sound(self):
        print("Cow says moo")


def make_sound(animal_object):
    animal_object.sound()


make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


n1 = Number(10)
n2 = Number(20)
result = n1 + n2
print("Sum:", result.value)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Python does not support traditional method overloading by defining
# the same method several times. We can use a default argument instead.

class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c


cal = Calculator()
print(cal.add(5))
print(cal.add(5, 10))
print(cal.add(5, 10, 15))
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


try:
    computer = Computer()
except TypeError as error:
    print("Error:", error)

# We cannot create an object until the abstract method is implemented.
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing")


laptop = Laptop()
desktop = Desktop()
laptop.process()
desktop.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod


class Car(ABC):
    def start_engine(self):
        print("Engine started")

    @abstractmethod
    def drive(self):
        pass


class MyCar(Car):
    def drive(self):
        print("Car is moving")


car = MyCar()
car.start_engine()
car.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Credit Card")


class UPIPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI")


payment1 = CreditCardPayment()
payment2 = UPIPayment()

payment1.pay(2500)
payment2.pay(1500)
```
